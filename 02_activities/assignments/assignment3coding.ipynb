```python
##Attendance at Drop-in Programs, no data for fall 2015 so removed it from the set for this visualization
year_Str=['2009 Act.',
'2010 Act.',
'2011 Act.',
'2012 Act.',
'2013 Act.',
'2014 Act.']


winter= [828642, 
 631802, 
 916249, 
 1186753, 
 973880, 
 1096461] 
spring=[ 501121, 
 406266, 
 517320, 
 750621, 
802679,
664922]
summer=[ 1100342,
 2225439, 
 2015070, 
 2067598, 
 1906097, 
 1753995 ]
fall=[ 622001, 
 947554, 
 861279, 
 959051, 
 1124760, 
 1040860 ]
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
fig, ax =plt.subplots(figsize=(10,5))

bottom = np.zeros(len(year_Str))##setting up array 'bottom' with 0 in the length of year, so we can stack 
                                #the seasons in the bar chart for each year
ax.bar(year_Str, winter, bottom=bottom, label="Winter",color='blue')##plot winter bar for each year with label "winter" and color blue,
bottom += np.array(winter) ##adding winter to bottom so the next bar can stack on top, repeat same process 
                            #for the other seasons

ax.bar(year_Str, spring, bottom=bottom, label="Spring",color='yellow')
bottom += np.array(spring)

ax.bar(year_Str, summer, bottom=bottom, label="Summer",color='r')
bottom += np.array(summer)

ax.bar(year_Str, fall, bottom=bottom, label="Fall",color='orange')
ax.set_xlabel("Year") ##x axis label
ax.set_ylabel('Attendance')#y axis label
ax.legend(loc='best')#setting legend to find auto best placement
ax.set_title("Attendance at drop-in programs")#set title
plt.savefig('/Users/spenserross/Desktop/Data Science/DSI Course/visualization/visualization/02_activities/assignments.png',dpi=300)
```
