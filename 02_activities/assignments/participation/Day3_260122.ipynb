```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613) 
x=np.arange(50)
y1=np.random.randint(0,100,50)
y2=np.random.randint(0,100,50)
```

```python
fig, ax =plt.subplots(figsize=(5,3))
ax.plot(x,y1, label='person 1')
ax.plot(x,y2,label='person 2')
ax.legend(loc='lower right')
plt.show()
```

```python
fig, ax =plt.subplots(figsize=(5,3))
ax.plot(x,y1, label='person 1')
ax.plot(x,y2,label='person 2')
ax.legend(loc='lower right', 
          frameon=True, #add frame around legend,
          fontsize=8,
          ncol=2,
          shadow=True)
plt.show()
```

```python
fig, ax =plt.subplots(figsize=(5,3))
ax.plot(x,y1, label='person 1')
ax.plot(x,y2,label='person 2')
ax.legend(loc='upper left', #######attaching upper left to the 1,1 point of fig
          bbox_to_anchor =(1,1))
ax.text(10,95, "This value is important",ha='center',size=5,color='k')
```

```python
fig,ax=plt.subplots()
ax.axis([0,10,0,10])
ax.text(1,5,". Data:(1,5)", transform=ax.transData)
ax.text(0.5,0.1,". Axes: (0.5, 1)", transform = ax.transAxes)
ax.text(0.5,0.1,". Figure:(0.2,0.2)", transform=fig.transFigure)
```

```python
fig, ax =plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label='person 1')
ax.scatter(x,y2,label='person 2')
ax.legend(loc='upper left')

ax.annotate('this is important',xy=(10,10),xytext=(20,94),arrowprops=dict(facecolor='black',arrowstyle='fancy'),color='hotpink')
```

```python
fig, ax =plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label='person 1')
ax.scatter(x,y2,label='person 2')
ax.legend(loc='upper left')
# ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_formatter(plt.NullFormatter())
```
