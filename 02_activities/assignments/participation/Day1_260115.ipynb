```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy 
import PIL
import requests
```

```python
np.random.seed(613)#set seed for reproducibility
x=np.arange(50) 
y=np.random.randint(0,100,50)##random number
```

```python
fig,ax=plt.subplots(figsize=(5,3))##set up fig and axes
# ax.scatter(x,y)#scatter plot
# ax.bar(x,y)#bar chart
# ax.plot(x,y) #line plot
ax.hist(y)###plot histogram
ax.hist(y,edgecolor='k')

```

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y)
ax.set_title('total growth over time')#make title

ax.set_ylabel('total growth')#ylabel
ax.set_xlabel('years since start')#xlabel
fig.tight_layout()#fix spacing
```

```python
#dicts for properties of plot labels
font1={'family':'sans-serif','color':'blue','size':20}
font2={'family':'monospace','color':'green','size':14}
```

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y)
ax.set_title('Total growth over time',fontdict=font1)
ax.set_ylabel('total growth',fontdict=font2)
ax.set_xlabel('years since start',fontdict=font2)
fig.tight_layout()
```

```python
fig,ax=plt.subplots(figsize=(5,3))##altering marker and colors
ax.scatter(x,
           y,
           marker='*',
           color='indigo')
```

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,
           y,
           color='indigo',
           marker='*',
           linestyle='--',
           linewidth=2,
           )
```

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,
           y,
           color='#7425b9',
           marker='*',
           linestyle='--',
           linewidth=2,
           )
```

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,
           y,
           color='#7425b9',
           marker='*',
           markersize=12,
           linestyle='--',
           linewidth=2,
           markeredgecolor='#fa9359',
           markerfacecolor='#000000',
           )
```
