```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x1=np.arange(50)
y1=np.random.randint(0,75,50)
x2=np.array(['Luffy', "Zoro", "Nami", "usopp", "Sanji"])
y2=np.array ([110,180,240,99,220])
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,nrows=1,figsize=(7,3))
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,nrows=1,figsize=(7,3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
ax1.annotate('this is important',xy=(10,65),xytext=(30,65),arrowprops=dict(facecolor='black'))
```

```python
fig,axes=plt.subplot_mosaic([['ax1','ax3'],['ax2','ax3']],figsize=(7,4),layout='constrained')
axes['ax1'].scatter(x1,y1)
axes['ax2'].bar(x2,y2)
axes['ax3'].plot(x1,y1)
axes['ax1'].set_xlabel(' a big label', fontsize =18)
axes['ax2'].set_xlabel(' another label', fontsize =18)
axes['ax3'].set_xlabel(' label 2 fast 2 furious', fontsize =18)
# fig.tight_layout()
```

```python

x=np.array(['Luffy', "Zoro", "Nami", "usopp", "Sanji"])
y1=np.array ([110,180,240,99,220])
y2=np.array([170,100,90,120,50])

fig,ax=plt.subplots(figsize=(7,3))
ax.bar(x,y1,color='indigo')
ax.plot(x,y2,color='r')
```

```python
y2_sd=np.std(y2)
fig,ax=plt.subplots(figsize=(7,3))
# ax.plot(x,y2,color='r')
ax.errorbar(x,#####x values
            y2,####y values
            yerr=y2_sd,####std
            color='r',
            fmt='o-',
            ecolor='b',
            elinewidth=4,
            capsize=6,
            capthick=4,
            errorevery=2)
```

```python
from io import BytesIO
from PIL import Image
```

```python
url ="https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"
headers={"User-Agent": "Mozilla/5.0","Accept": "image/*,*/*;q=0.8","Referer": 'https://en.wikipedia.org/'}###add headers to help extract image, cuz wiki blocks axcess
response=requests.get(url,headers=headers)#request to get image from url
image_file=BytesIO(response.content)##download image
image=Image.open(image_file)#open image
image
```

```python
x=np.array(['Luffy', "Zoro", "Nami", "usopp", "Sanji"])
y1=np.array ([110,180,240,99,220])
y2=np.array([170,100,90,120,50])
fig,ax=plt.subplots(figsize=(7,3))
ax.plot(x,y2,color='r')
ax_image=fig.add_axes([0.1,0.11,0.15,0.35])#xcoord,ycoord,image width,image height///coords are on fig not ax
ax_image.imshow(image)
ax_image.axis('off')
path='/Users/spenserross/Desktop/Data Science/DSI Course/visualization'
filename='/fig1a.png'
plt.savefig(path+filename,dpi=300) 
```
